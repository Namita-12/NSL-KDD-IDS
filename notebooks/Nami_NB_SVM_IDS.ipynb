```python
import numpy as np
import pandas as pd
import joblib

from sklearn.naive_bayes import GaussianNB
from sklearn.svm import LinearSVC
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)

print("Imports successful.")
```

```python
processed_path = "/content/drive/MyDrive/IDS_Project/output/processed.pkl"

d = joblib.load(processed_path)

X_train = d["X_train"]
X_test = d["X_test"]

y_train = d["y_train"]
y_test = d["y_test"]

class_names = d["class_names"]
feature_names = d["feature_names"]

print("Processed data loaded successfully.")
```

```python
print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)

print("y_train shape:", y_train.shape)
print("y_test shape :", y_test.shape)

print("\nClass names:")
print(class_names)

print("\nNumber of features:", len(feature_names))

print("\nData types:")
print("X_train:", X_train.dtype)
print("X_test :", X_test.dtype)
print("y_train:", y_train.dtype)
print("y_test :", y_test.dtype)
```

```python
# ============================================
# Gaussian Naive Bayes
# ============================================

gnb = GaussianNB()

# Train ONLY on the training data
gnb.fit(X_train, y_train)

print("Gaussian Naive Bayes training completed.")
```

```python
# Generate predictions on the official KDDTest+ set
y_pred_gnb = gnb.predict(X_test)

print("Prediction completed.")
print("Number of predictions:", len(y_pred_gnb))
```

```python
# ============================================
# Overall Evaluation
# ============================================

gnb_accuracy = accuracy_score(y_test, y_pred_gnb)

gnb_macro_precision = precision_score(
    y_test,
    y_pred_gnb,
    average="macro",
    zero_division=0
)

gnb_macro_recall = recall_score(
    y_test,
    y_pred_gnb,
    average="macro",
    zero_division=0
)

gnb_macro_f1 = f1_score(
    y_test,
    y_pred_gnb,
    average="macro",
    zero_division=0
)

print("Gaussian Naive Bayes Results")
print("-" * 35)
print(f"Accuracy          : {gnb_accuracy:.4f}")
print(f"Macro Precision   : {gnb_macro_precision:.4f}")
print(f"Macro Recall      : {gnb_macro_recall:.4f}")
print(f"Macro F1          : {gnb_macro_f1:.4f}")
```

```python
# ============================================
# Per-Class Evaluation
# ============================================

print("Per-Class Classification Report")
print("-" * 50)

print(
    classification_report(
        y_test,
        y_pred_gnb,
        target_names=class_names,
        digits=4,
        zero_division=0
    )
)
```

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import LinearSVC

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=RANDOM_STATE
)

scoring = {
    "accuracy": "accuracy",
    "macro_precision": "precision_macro",
    "macro_recall": "recall_macro",
    "macro_f1": "f1_macro"
}

models = {
    "Decision Tree - Gini": DecisionTreeClassifier(
        criterion="gini",
        random_state=RANDOM_STATE
    ),

    "Decision Tree - Entropy": DecisionTreeClassifier(
        criterion="entropy",
        random_state=RANDOM_STATE
    ),

    "Gaussian Naive Bayes": GaussianNB(),

    "LinearSVC": LinearSVC(
        random_state=RANDOM_STATE,
        C=0.1,
        dual=False,
        tol=1e-3,
        max_iter=20000
    ),

    "Balanced LinearSVC": LinearSVC(
        random_state=RANDOM_STATE,
        C=0.1,
        dual=False,
        tol=1e-3,
        max_iter=20000,
        class_weight="balanced"
    )
}

print("5-fold CV setup ready.")
print("Number of folds:", cv.n_splits)
print("Models:", len(models))
```

```python
cv_results = []
cv_fold_results = {}

for name, model in models.items():

    print(f"\nRunning CV for: {name}")

    scores = cross_validate(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        n_jobs=-1,
        return_train_score=False
    )

    cv_fold_results[name] = scores

    cv_results.append({
        "Model": name,

        "Accuracy Mean": scores["test_accuracy"].mean(),
        "Accuracy Std": scores["test_accuracy"].std(),

        "Macro Precision Mean": scores["test_macro_precision"].mean(),
        "Macro Precision Std": scores["test_macro_precision"].std(),

        "Macro Recall Mean": scores["test_macro_recall"].mean(),
        "Macro Recall Std": scores["test_macro_recall"].std(),

        "Macro F1 Mean": scores["test_macro_f1"].mean(),
        "Macro F1 Std": scores["test_macro_f1"].std()
    })

    print(
        f"Macro F1: "
        f"{scores['test_macro_f1'].mean():.4f} "
        f"+/- {scores['test_macro_f1'].std():.4f}"
    )

print("\nAll 5-fold CV experiments completed.")
```

```python
import time
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import LinearSVC

latency_models = {
    "Decision Tree - Gini": DecisionTreeClassifier(
        criterion="gini",
        random_state=RANDOM_STATE
    ),

    "Decision Tree - Entropy": DecisionTreeClassifier(
        criterion="entropy",
        random_state=RANDOM_STATE
    ),

    "Gaussian Naive Bayes": GaussianNB(),

    "LinearSVC": LinearSVC(
        random_state=RANDOM_STATE,
        C=0.1,
        dual=False,
        tol=1e-3,
        max_iter=20000
    ),

    "Balanced LinearSVC": LinearSVC(
        random_state=RANDOM_STATE,
        C=0.1,
        dual=False,
        tol=1e-3,
        max_iter=20000,
        class_weight="balanced"
    )
}

# Use exactly 10,000 test connections
X_latency = X_test[:10000]

print("Latency benchmark samples:", X_latency.shape)
```

```python
trained_latency_models = {}

for name, model in latency_models.items():

    print(f"Training: {name}")

    model.fit(X_train, y_train)

    trained_latency_models[name] = model

print("\nAll latency benchmark models trained.")
```

```python
latency_results = []

for name, model in trained_latency_models.items():

    # Warm-up prediction
    model.predict(X_latency[:100])

    times = []

    for _ in range(5):
        start = time.perf_counter()

        model.predict(X_latency)

        end = time.perf_counter()

        times.append(end - start)

    avg_time = np.mean(times)

    connections_per_ms = 10000 / (avg_time * 1000)

    latency_results.append({
        "Model": name,
        "Average Prediction Time (s)": avg_time,
        "Time per 10,000 Connections (ms)": avg_time * 1000,
        "Connections per ms": connections_per_ms
    })

latency_df = pd.DataFrame(latency_results)

display(latency_df)
```
